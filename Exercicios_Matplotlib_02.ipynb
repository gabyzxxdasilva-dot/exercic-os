### Exercício 1: Gráfico de Temperaturas Mensais

Objetivo: Praticar gráfico de linha com múltiplas séries

Crie um gráfico de linha que mostre a temperatura média de duas cidades diferentes ao longo do ano.

```python
# Dados fornecidos
meses = ['Jan', 'Fev', 'Mar', 'Abr', 'Mai', 'Jun', 'Jul', 'Ago', 'Set', 'Out', 'Nov', 'Dez']
temperatura_cidade_a = [28, 27, 26, 24, 21, 19, 18, 19, 21, 23, 25, 27]
temperatura_cidade_b = [32, 31, 29, 26, 23, 21, 20, 21, 23, 26, 28, 30]

# SEU CÓDIGO AQUI:
# 1. Crie um gráfico de linha com duas séries (uma para cada cidade)
# 2. Use cores diferentes e legenda
# 3. Adicione título e rótulos nos eixos
# 4. Adicione grade para melhor leitura

import matplotlib.pyplot as plt

meses = ['Jan', 'Fev', 'Mar', 'Abr', 'Mai', 'Jun', 'Jul', 'Ago', 'Set', 'Out', 'Nov', 'Dez']

temperatura_cidade_a = [28, 27, 26, 24, 21, 19, 18, 19, 21, 23, 25, 27]

temperatura_cidade_b = [32, 31, 29, 26, 23, 21, 20, 21, 23, 26, 28, 30]

plt.plot(meses, temperatura_cidade_a, label='Cidade A', color='blue')
plt.plot(meses, temperatura_cidade_b, label='Cidade B', color='red')

plt.title('Temperatura das Cidades durante o Ano')
plt.xlabel('Meses')
plt.ylabel('Temperatura (°C)')

plt.legend()
plt.grid(True)

plt.show()
```

### Exercício 2: Popularidade de Linguagens de Programação

**Objetivo**: Praticar gráfico de barras horizontais

Crie um gráfico de barras horizontais mostrando a popularidade de diferentes linguagens de programação.

```python
# Dados fornecidos
linguagens = ['Python', 'JavaScript', 'Java', 'C#', 'PHP', 'C++']
popularidade = [45, 40, 35, 25, 20, 18]  # em porcentagem

# SEU CÓDIGO AQUI:
# 1. Crie um gráfico de barras HORIZONTAIS
# 2. Use uma paleta de cores gradiente (mais popular = cor mais forte)
# 3. Adicione os valores no final de cada barra
# 4. Inverta a ordem para mostrar a mais popular no topo

import matplotlib.pyplot as plt
import numpy as np

linguagens = ['Python', 'JavaScript', 'Java', 'C#', 'PHP', 'C++']
popularidade = [45, 40, 35, 25, 20, 18]

cores = plt.cm.Blues(np.linspace(1, 0.4, len(linguagens)))

plt.barh(linguagens, popularidade, color=cores)

for i, valor in enumerate(popularidade):
    plt.text(valor + 1, i, f'{valor}%', va='center')

plt.gca().invert_yaxis()

plt.title('Popularidade das Linguagens')
plt.xlabel('Popularidade (%)')
plt.ylabel('Linguagens')

plt.show()
```

### Exercício 3: Distribuição de Notas da Turma

**Objetivo**: Praticar histograma e personalização

Crie um histograma mostrando a distribuição das notas de uma turma em uma prova.

```python
import numpy as np

# Gerando dados aleatórios de notas (0 a 10)
np.random.seed(123)
notas = np.random.normal(7.0, 1.5, 200)  # média 7, desvio padrão 1.5
notas = np.clip(notas, 0, 10)  # limitando entre 0 e 10

# SEU CÓDIGO AQUI:
# 1. Crie um histograma com 15 bins (caixas)
# 2. Adicione uma linha vertical na média
# 3. Customize cores e transparência
# 4. Adicione título e rótulos descritivos

import numpy as np
import matplotlib.pyplot as plt

np.random.seed(123)
notas = np.random.normal(7.0, 1.5, 200)
notas = np.clip(notas, 0, 10)

media = np.mean(notas)

plt.hist(notas, bins=15, color='skyblue', alpha=0.7, edgecolor='black')
plt.axvline(media, color='red', linestyle='--', linewidth=2, label=f'Média: {media:.2f}')

plt.title('Distribuição das Notas dos Alunos')
plt.xlabel('Notas')
plt.ylabel('Quantidade de Alunos')
plt.legend()

plt.show()
```

### Exercício 4: Preferência por Esportes

**Objetivo**: Praticar gráfico de pizza (pie chart) com customização

Crie um gráfico de pizza mostrando a preferência por esportes em uma pesquisa.

```python
# Dados fornecidos
esportes = ['Futebol', 'Vôlei', 'Basquete', 'Natação', 'Tênis', 'Outros']
preferencia = [35, 20, 15, 12, 10, 8]  # em porcentagem
cores = ['#FF6B6B', '#4ECDC4', '#45B7D1', '#96CEB4', '#FDCB6E', '#BDC3C7']

# SEU CÓDIGO AQUI:
# 1. Crie um gráfico de pizza
# 2. Use as cores fornecidas
# 3. Exploda o pedaço do esporte mais popular
# 4. Adicione porcentagens e legenda
# 5. Garanta que o gráfico seja circular (aspect ratio igual)
import matplotlib.pyplot as plt

esportes = ['Futebol', 'Vôlei', 'Basquete', 'Natação', 'Tênis', 'Outros']
preferencia = [35, 20, 15, 12, 10, 8]

cores = ['#FF6B6B', '#4ECDC4', '#45B7D1', '#96CEB4', '#FDCB6E', '#BDC3C7']

explode = [0.1, 0, 0, 0, 0, 0]

plt.pie(
    preferencia,
    labels=esportes,
    colors=cores,
    explode=explode,
    autopct='%1.1f%%'
)

plt.title('Preferência por Esportes')
plt.legend(esportes)
plt.axis('equal')

plt.show()
```

### Exercício 5: Dashboard de Métricas

**Objetivo**: Praticar subplots múltiplos

Crie uma figura com 4 subplots diferentes mostrando diversos tipos de visualização.

```python
import numpy as np

# Dados para os gráficos
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)

categorias = ['A', 'B', 'C', 'D']
valores1 = [23, 45, 56, 78]
valores2 = [45, 34, 30, 50]

dados_hist = np.random.normal(0, 1, 1000)

# SEU CÓDIGO AQUI:
# Crie uma figura com 2x2 subplots (4 gráficos no total):

# Gráfico 1 (superior esquerdo): Linha - sen(x) e cos(x)
# Gráfico 2 (superior direito): Barras - valores1 e valores2 (barras lado a lado)
# Gráfico 3 (inferior esquerdo): Histograma - dados_hist
# Gráfico 4 (inferior direito): Dispersão - x vs y1

# Dicas:
# - Use plt.subplots(2, 2, figsize=(12, 8))
# - Customize cada gráfico individualmente
# - Use plt.tight_layout() para melhor espaçamento

import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)

categorias = ['A', 'B', 'C', 'D']
valores1 = [23, 45, 56, 78]
valores2 = [45, 34, 30, 50]

dados_hist = np.random.normal(0, 1, 1000)

fig, axs = plt.subplots(2, 2, figsize=(12, 8))

axs[0, 0].plot(x, y1, label='sen(x)')
axs[0, 0].plot(x, y2, label='cos(x)')
axs[0, 0].set_title('Sen(x) e Cos(x)')
axs[0, 0].set_xlabel('X')
axs[0, 0].set_ylabel('Y')
axs[0, 0].legend()
axs[0, 0].grid(True)

largura = 0.35
posicoes = np.arange(len(categorias))

axs[0, 1].bar(posicoes - largura/2, valores1, largura, label='Valores 1')
axs[0, 1].bar(posicoes + largura/2, valores2, largura, label='Valores 2')
axs[0, 1].set_title('Comparação de Valores')
axs[0, 1].set_xticks(posicoes)
axs[0, 1].set_xticklabels(categorias)
axs[0, 1].legend()

axs[1, 0].hist(dados_hist, bins=30, edgecolor='black')
axs[1, 0].set_title('Histograma')
axs[1, 0].set_xlabel('Valores')
axs[1, 0].set_ylabel('Frequência')

axs[1, 1].scatter(x, y1)
axs[1, 1].set_title('Dispersão: X vs Sen(X)')
axs[1, 1].set_xlabel('X')
axs[1, 1].set_ylabel('Sen(X)')
axs[1, 1].grid(True)

plt.tight_layout()
plt.show()
```
